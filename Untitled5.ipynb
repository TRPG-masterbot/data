```python
with open("rules.jsonl", encoding="utf-8") as f:
    chunk = json.loads(f.readline())
# with open(...): 파일을 열고, 블록이 끝나면 자동으로 닫음
# encoding="utf-8": 한글·특수문자가 깨지지 않게함
# f.readline(): 첫 줄 하나만 읽어요. jsonl은 한 줄이 JSON 하나(청크 하나)라서, 한 줄 = 청크 1개
# json.loads(...): 문자열 → 딕셔너리
def embed(text):
    r = requests.post(
        "http://localhost:11434/api/embed",
        json={"model": "bge-m3", "input": text},
        timeout=60,
    )
    r.raise_for_status()
    return r.json()["embeddings"][0]
# post: 서버에 데이터를 보내는 요청
# r.raise_for_status() : 응답이 에러(404, 500 등)면 여기서 바로 예외를 던짐
#[0]: 그중 첫 번째 벡터만 꺼냄 (숫자 1024개의 list)
vec = embed(chunk["text"])
print(chunk["meta"]["title"], len(vec))
```

```python
OLLAMA_URL = "http://localhost:11434"
EMBED_MODEL = "bge-m3"
BATCH = 32
DATA = Path("/content/drive/MyDrive/KT 기본프로젝트")
def embed_batch(texts):
    r = requests.post(
        f"{OLLAMA_URL}/api/embed",
        json={"model": EMBED_MODEL, "input": texts},
        timeout=300,
    )
    r.raise_for_status()
    return r.json()["embeddings"]

def load_chunks(path):
    with open(path, encoding="utf-8") as f:
        return [json.loads(line) for line in f]

def build_embeddings(chunks):
    vecs = []
    for i in tqdm(range(0, len(chunks), BATCH)):
        batch = [c["text"] for c in chunks[i:i + BATCH]]
        vecs.extend(embed_batch(batch))
    arr = np.array(vecs, dtype="float32")
    assert arr.shape == (len(chunks), 1024), f"shape 이상: {arr.shape}"
    return arr

chunks = load_chunks(DATA / "rules.jsonl")
cache = DATA / f"rules_{EMBED_MODEL}.npy"

if cache.exists():
    arr = np.load(cache)
    assert len(arr) == len(chunks)
else:
    arr = build_embeddings(chunks)
    np.save(cache, arr)

print(arr.shape)
# embed_batch: 리스트를 한 번에 보냄
#chunks[i:i + BATCH]: 32개씩 잘라서 보냄
# vecs.extend(...): 리스트를 이어 붙임
# assert: 개수나 차원이 틀리면 바로 멈춤
```

```python
!pip install -q chromadb
import chromadb

CHROMA_PATH = "/content/chroma_db"
COLLECTION = f"rules_{EMBED_MODEL}"

def clean_meta(meta):
    return {k: ("" if v is None else v) for k, v in meta.items()}

def build_collection(chunks, arr):
    ids = [c["id"] for c in chunks]
    assert len(set(ids)) == len(ids), "id 중복"

    client = chromadb.PersistentClient(path=CHROMA_PATH)
    try:
        client.delete_collection(COLLECTION)
    except Exception:
        pass
    col = client.create_collection(
        COLLECTION, metadata={"hnsw:space": "cosine"}
    )

    for i in range(0, len(chunks), 500):
        part = chunks[i:i + 500]
        col.add(
            ids=[c["id"] for c in part],
            documents=[c["text"] for c in part],
            embeddings=arr[i:i + 500].tolist(),
            metadatas=[clean_meta(c["meta"]) for c in part],
        )

    assert col.count() == len(chunks)
    return col

col = build_collection(chunks, arr)
print(col.count())
#hnsw:space: cosine: 거리 기준
```

```python
def search(query, k=5):
    res = col.query(
        query_embeddings=[embed(query)],
        n_results=k,
        include=["metadatas", "distances"],
    )
    for m, d in zip(res["metadatas"][0], res["distances"][0]):
        print(round(d, 3), m["title"])

search("How does grappling work?")
```

```python
golden = json.load(open(DATA / "golden.json", encoding="utf-8"))

def is_hit(title, want):
    wants = [want] if isinstance(want, str) else want
    return any(w in title for w in wants)

def recall_at_k(golden, ks=(1, 3, 5)):
    hits = {k: 0 for k in ks}
    misses = []
    for g in golden:
        res = col.query(
            query_embeddings=[embed(g["q"])],
            n_results=max(ks),
            include=["metadatas"],
        )
        titles = [m["title"] for m in res["metadatas"][0]]
        # 정답이 처음 나온 순위
        rank = next(
            (i + 1 for i, t in enumerate(titles) if is_hit(t, g["want"])), None
        )
        for k in ks:
            if rank is not None and rank <= k:
                hits[k] += 1
        if rank is None or rank > 3:
            misses.append((g["q"], g["want"], titles[:3]))
    for k in ks:                    # 3등 안에 못 든 문항을 정답과 상위 3개 제목과 함께 보여줌
        print(f"Recall@{k}: {hits[k]}/{len(golden)} = {hits[k]/len(golden):.0%}")
    return misses

messes = recall_at_k(golden)
for q, want, top in messes:
    print("-", q, "| 정답:", want, "| 상위3:", top)
# want가 문자열이면 그 제목, 리스트면 그중 하나만 맞아도 정답
# w in title: 부분 문자열 일치 # "Grappl" -> Grappling, Grappled
# rank: 정답이 처음 나온 순위. 순위가 k 이하면 Recall@k 성공
# 검색은 한 번만 하고(n_results=5) 1, 3, 5를 같이 계산
```
